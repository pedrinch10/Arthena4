```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Columnas relevantes excluyendo IDs o URLs largas (basado en tu describe)
columnas_plot = ['pactivos', 'labtitular', 'labcomercializador', 'dosis', 'forma_farmaceutica_simplificada', 'vias_administracion']

# Ajustamos el tamaño y la cuadrícula de los gráficos (ej. 2 filas, 3 columnas)
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(18, 10))
axes = axes.flatten()

for i, col in enumerate(columnas_plot):
    # Calculamos las frecuencias absolutas
    conteos = df[col].value_counts()
    
    # Extraemos el Top 5
    top_5 = conteos.head(5)
    
    # Sumamos el resto de las categorías bajo la etiqueta "Otros"
    if len(conteos) > 5:
        otros = pd.Series([conteos.iloc[5:].sum()], index=['Otros'])
        datos_plot = pd.concat([top_5, otros])
    else:
        datos_plot = top_5
        
    # Dibujamos el gráfico de tarta
    axes[i].pie(
        datos_plot, 
        labels=datos_plot.index, 
        autopct='%1.1f%%', 
        startangle=140,
        colors=plt.cm.Paired(np.arange(len(datos_plot))) # Opcional: paleta de colores
    )
    axes[i].set_title(f'Top 5 - {col}')

# Ocultar el último eje (ya que tenemos 5 columnas y 6 espacios en el grid)
axes[-1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Añadimos 'vias_administracion' a la lista
columnas_plot = [
    'pactivos', 
    'labtitular', 
    'labcomercializador', 
    'dosis', 
    'forma_farmaceutica_simplificada', 
    'vias_administracion'
]

# Mantenemos el grid de 3 filas x 2 columnas (perfecto para 6 gráficos)
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(18, 16))
axes = axes.flatten()

for i, col in enumerate(columnas_plot):
    # Extraemos el Top 10
    top_10 = df[col].value_counts().head(10)
    
    # Invertimos el orden para que la categoría con más valores quede arriba
    categorias = top_10.index[::-1]
    frecuencias = top_10.values[::-1]
    
    # Creamos el gráfico de barras horizontales
    axes[i].barh(categorias, frecuencias, color='steelblue', edgecolor='black')
    axes[i].set_title(f'Top 10 - {col}', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Frecuencia')
    
    # Añadimos el valor exacto al lado de cada barra
    for index, value in enumerate(frecuencias):
        axes[i].text(value, index, f' {value}', va='center')

# Ajustamos el espaciado para que no se solapen los nombres largos
plt.tight_layout()
plt.show()
```
